```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/raw/train.csv")
```

```python
corr = df.corr(numeric_only=True)
corr
```

```python
corr["SalePrice"].sort_values(ascending=False)
```

```python
"""Do bigger house always cost more?"""
plt.figure(figsize=(7,5))

plt.scatter(df["GrLivArea"], df["SalePrice"])

plt.xlabel("Living Area")

plt.ylabel("Sale Price")

plt.title("Living Area vs Sale Price")

plt.show()
```

```python
"""Now look at overall quality
   The better quality tend to cost more?"""
plt.figure(figsize=(7,5))

plt.scatter(df["OverallQual"], df["SalePrice"])

plt.xlabel("Overall Quality")

plt.ylabel("Sale Price")

plt.title("Overall Quality vs Sale Price")

plt.show()
```

### Correlation

The correlation value shows the Overall Quality is the main factor affect the saleprice, from the scatter plot we can see the trend the better overall quality the higher saleprice.
