```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/raw/train.csv")
```

```python
"""try to create some new variables"""
df["PricePerSqFt"] = df["SalePrice"] / df["GrLivArea"]
df.PricePerSqFt
```

```python
"""how many bathroom above ground"""
df["BathAbvGr"] = df["FullBath"] + df["HalfBath"] - df["BsmtFullBath"] - df["BsmtHalfBath"]
df.BathAbvGr
```

```python
"""Get the houses age when it sold"""
df["HouseAge"] = df["YrSold"] - df["YearBuilt"]
df.HouseAge
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["HouseAge"], bins=25)

plt.title("Distribution of House Age")

plt.xlabel("Years")

plt.ylabel("Frequency")

plt.show()
```
